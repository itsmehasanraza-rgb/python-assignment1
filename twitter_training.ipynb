```python
import pandas as pd
import numpy as np

from google.colab import files

uploaded = files.upload()

df = pd.read_csv(
    'twitter_training.csv',
    header=None,
    names=['Tweet_ID', 'Entity', 'Sentiment', 'Tweet_Text']
)

print("Dataset Loaded Successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

```python
# Display first 5 rows
df.head()

# Dataset information
df.info()

# Check missing values
print(df.isnull().sum())

# Check sentiment distribution
print(df['Sentiment'].value_counts())
```

```python
# Step 3: Handling Missing Values

# Check missing values
print("Missing values before cleaning:")
print(df.isnull().sum())

# Remove empty text values
df['Tweet_Text'] = df['Tweet_Text'].replace(
    r'^\s*$', np.nan, regex=True
)

# Remove rows where Tweet_Text is missing
df.dropna(subset=['Tweet_Text'], inplace=True)

# Reset index
df.reset_index(drop=True, inplace=True)

print("\nMissing values after cleaning:")
print(df.isnull().sum())

print("\nNew Dataset Shape:", df.shape)
```

```python
# Step 4: Removing Duplicate Records

# Count duplicate rows
print("Duplicate rows before removal:", df.duplicated().sum())

# Remove duplicate rows
df.drop_duplicates(inplace=True)

# Reset index
df.reset_index(drop=True, inplace=True)

print("Duplicate rows after removal:", df.duplicated().sum())

print("Final Dataset Shape:", df.shape)
```

```python

# Step 5: Text Cleaning

import re

def clean_text(text):

    # Remove URLs
    text = re.sub(r'http\S+|www\S+|pic\.twitter\.com\s*/?\s*\S*', '', str(text))

    # Remove Twitter mentions
    text = text.replace('@', '')

    # Remove HTML tags
    text = re.sub(r'<.*?>', '', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    text = text.replace('@', '')

    return text

# Create a new column for cleaned text
df['Cleaned_Text'] = df['Tweet_Text'].apply(clean_text)

# Display original and cleaned tweets
df[['Tweet_Text', 'Cleaned_Text']].head(10)
```

```python
# Step 6: Lowercase Conversion

df['Cleaned_Text'] = df['Cleaned_Text'].str.lower()

# Display results
df[['Cleaned_Text']].head(10)
```

```python
# Step 7: Removing Punctuation

import string

def remove_punctuation(text):
    return text.translate(
        str.maketrans('', '', string.punctuation)
    )

df['Cleaned_Text'] = df['Cleaned_Text'].apply(remove_punctuation)

# Display results
df[['Cleaned_Text']].head(10)
```

```python
# Step 8: Removing Numbers

df['Cleaned_Text'] = df['Cleaned_Text'].str.replace(
    r'\d+', '', regex=True
)

# Display results
df[['Cleaned_Text']].head(10)
```

```python
# Step 9: Tokenization

import nltk

nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize

# Convert text into tokens
df['Tokens'] = df['Cleaned_Text'].apply(word_tokenize)

# Display results
df[['Cleaned_Text', 'Tokens']].head(10)
```

```python
# Step 10: Stopwords Removal

import nltk

nltk.download('stopwords')

from nltk.corpus import stopwords

# Get English stopwords
stop_words = set(stopwords.words('english'))

# Remove stopwords
df['Filtered_Tokens'] = df['Tokens'].apply(
    lambda tokens: [word for word in tokens if word not in stop_words]
)

# Display results
df[['Tokens', 'Filtered_Tokens']].head(10)
```

```python
# Step 11: Stemming

from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

# Apply stemming
df['Stemmed_Tokens'] = df['Filtered_Tokens'].apply(
    lambda tokens: [stemmer.stem(word) for word in tokens]
)

# Display results
df[['Filtered_Tokens', 'Stemmed_Tokens']].head(10)
```

```python
# Step 12: Lemmatization

import nltk

nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger_eng')

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

# Apply lemmatization
df['Lemmatized_Tokens'] = df['Filtered_Tokens'].apply(
    lambda tokens: [lemmatizer.lemmatize(word) for word in tokens]
)

# Display results
df[['Filtered_Tokens', 'Stemmed_Tokens', 'Lemmatized_Tokens']].head(10)
```

```python
# Step 13: Creating Final Processed Text

df['Final_Processed_Text'] = df['Lemmatized_Tokens'].apply(
    lambda tokens: ' '.join(tokens)
)

# Display original and processed tweets
df[['Tweet_Text', 'Final_Processed_Text']].head(10)
```

```python
# Save processed dataset as CSV

df.to_csv('twitter_preprocessed.csv', index=False)

print("Preprocessed dataset saved successfully!")
```
