# User se marks lo aur grade do:

# 90+ → A
# 80+ → B
# 70+ → C
# 60+ → D
# warna → Fail

```python
###### sub1 = int(input("enter a number:"))
sub2 = int(input("enter a number:"))
sub3 = int(input("enter a number:"))
sub4 = int(input("enter a number:"))
sub5 = int(input("enter a number:"))
obtain_marks = sub1+sub2+sub3+sub4+sub5
pec = (obtain_marks/500)*100
grade = ""
if pec>=90:
    grade = "A+"
elif pec>=80:
    grade = "B"
elif pec>=70:
    grade = "c"
elif pec>=60:
    grade = "D"
else:
    grade = "Fail"

print(f"""
      total Marks =  500
      obtain Marks : {obtain_marks}
      percentage :  {pec}
      Grade  :       {grade}
      """)    
```

# User se month number lo:

# 12,1,2 → Winter
# 3,4,5 → Spring
# 6,7,8 → Summer
# warna → Autumn

```python
month = int(input("enter the month number"))
if month in [12, 1, 2]:
    print("winter")
elif month in [3, 4, 5]:
    print("Spring")
elif month in [6, 7, 8]:
    print("Summer")
elif month in [9, 10, 11]:
    print("Autumn")
else:
    print("Invalid number")
```

# Simple calculator:
# User se operator lo (+, -, *, /):

# → addition
# → subtraction
# → multiplication
# / → division
# warna → "Invalid operator"

```python
num1 = int(input("enter a number"))
num2 = int(input("enter a number"))
operator = (input("enter an operator from these: +, -, /, * "))
ans = 0
if operator == "+":
    ans = num1 + num2
elif operator == "-":
    ans = num1 - num2
elif operator == "/":
    ans = num1/num2    
elif operator == "*":
    ans = num1*num2
else:
    print("Invalid operator")


print(f"""
      ans : {ans}
      """)

    
```
